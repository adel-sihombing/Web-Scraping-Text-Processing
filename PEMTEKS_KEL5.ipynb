```python
import requests
from bs4 import BeautifulSoup

page = requests.get('https://github.com/')
bs = BeautifulSoup(page.text, 'html.parser')

bs
```

```python
text = bs.get_text(separator=' ', strip=True)

print(text[:2000])
```

```python
import re
```

```python
#lowercase
hasil_cf = str.lower(text)
print(hasil_cf)
```

```python
#remove punctuation
import string
punc = string.punctuation
def remove_punctuation(text):
    return text.translate(str.maketrans('', '', punc))

hasil_rp = remove_punctuation(text)
print(hasil_rp)
```

```python
#stopwords
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')

print(stopwords.words('english'))
print(stopwords.words('indonesian'))
```

```python
sw = set(stopwords.words('english'))
def remove_stopwords(text):

  return " ".join([word for word in str(text).split() if word not in sw])

hasil_st = remove_stopwords(text)
print(hasil_st)
```

```python
# Stemmer

import nltk
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

print(stemmer.stem('developers'))     # => develop
print(stemmer.stem('developing'))     # => develop
print(stemmer.stem('applications'))   # => applic
print(stemmer.stem('connected'))      # => connect

sentence = "The future of building happens together"

print(" ".join(
    stemmer.stem(word)
    for word in sentence.split()
))
```

```python
from nltk.stem import LancasterStemmer
ls = LancasterStemmer()
ls.stem('developers'), ls.stem('developing'), ls.stem('applications'), ls.stem('connected')
```

```python
from nltk.stem import SnowballStemmer
ss = SnowballStemmer("english")
ss.stem('student'), ss.stem('studying'), ss.stem('studies'), ss.stem('learning')
```

```python
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
wnl = WordNetLemmatizer()

# Lemmatize nouns
print(wnl.lemmatize('developers', 'n'))     # developer
print(wnl.lemmatize('applications', 'n'))   # application
# Lemmatize verbs
print(wnl.lemmatize('building', 'v'))       # build
print(wnl.lemmatize('happens', 'v'))        # happen
print(wnl.lemmatize('connected', 'v'))      # connect
```
